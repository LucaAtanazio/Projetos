```python
import pandas as pd
import random
import time
from IPython.display import clear_output, display, HTML
```

# Carregamento de .csv

```python
# 1. Carregar o arquivo CSV (substitua pelo nome real do seu arquivo se necessário)
# Certifique-se de que o arquivo esteja no mesmo diretório que o notebook.
# O pandas lê o arquivo e cria um DataFrame, que é uma estrutura de tabela.
try:
    # Ler o CSV, assumindo que há um cabeçalho e os dados estão em uma coluna
    df = pd.read_csv('watchlist_18-04-2026.csv', header=None, names=['Filme'])
    
    # Extrair os nomes dos filmes para uma lista, removendo possíveis valores vazios
    filmes = df['Filme'].dropna().tolist()
    
    # Se a primeira linha for um cabeçalho como 'title' ou 'filme', você pode querer
    # remover essa linha se ela não for o nome de um filme real.
    # filmes = filmes[1:] # Descomente essa linha se tiver cabeçalho e quiser pular a 1ª linha

except FileNotFoundError:
    print("Erro: O arquivo 'watchlist.csv' não foi encontrado.")
    print("Verifique se o arquivo está na mesma pasta que este notebook.")
    filmes = [] # Define filmes como lista vazia para evitar erros posteriores

# 2. Verificar se a lista de filmes está vazia
if not filmes:
    print("Sua watchlist está vazia ou o arquivo não pôde ser lido.")
else:
    # 3. Criar a animação e o sorteio
    
    # Número de iterações para a animação (quantos filmes "passarão")
    iteracoes = 30
    
    # Tempo de espera entre cada filme (segundos) - para um efeito de desaceleração
    # Começa rápido e vai ficando mais lento
    tempos_espera = [max(0.01, 0.2 - (i / iteracoes) * 0.19) for i in range(iteracoes)]

    print("Preparando para sortear...")
    time.sleep(1) # Uma pequena pausa antes de começar

    # Loop para a animação
    for i in range(iteracoes):
        # Limpa a saída anterior da célula para dar o efeito de animação
        clear_output(wait=True)
        
        # Escolhe aleatoriamente um filme (apenas para exibição momentânea)
        filme_temp = random.choice(filmes)
        
        # Exibe o filme com alguma formatação HTML simples (tamanho maior e centralizado)
        # O HTML precisa estar em uma única linha para o display() funcionar corretamente.
        html_content = f"<h2 style='text-align: center; color: #555;'>Sorteando: <br><span style='font-size: 1.5em; color: #333;'>{filme_temp}</span></h2>"
        display(HTML(html_content))
        
        # Espera o tempo definido para a iteração atual (vai ficando mais lento)
        time.sleep(tempos_espera[i])
```

# Sorteio do Filme

```python
# 4. Sorteio Final e Exibição com destaque
    clear_output(wait=True) # Limpa a última iteração da animação
    
    filme_sorteado = random.choice(filmes)
    
    # Exibe o filme sorteado com destaque usando HTML e CSS
    html_sorteado = f"""
    <div style='text-align: center; border: 2px solid #333; padding: 20px; background-color: #f9f9f9; border-radius: 10px; box-shadow: 0 4px 8px rgba(0,0,0,0.1);'>
        <h1 style='color: #222; margin-top: 0;'>🎥 Filme Sorteado! 🍿</h1>
        <hr style='border: 1px solid #ccc; width: 50%;'>
        <h2 style='color: #007bff; font-size: 2.5em; margin-bottom: 10px;'>{filme_sorteado}</h2>
        <p style='color: #555; font-style: italic;'>Aproveite sua sessão!</p>
    </div>
    """
    display(HTML(html_sorteado))
```
